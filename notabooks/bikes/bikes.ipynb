```python
df.dropna(inplace=True)
df.info()
```

```python
df[ 'started_at'] = pd.to_datetime(df[ 'started_at'])
df[ 'ended_at'] = pd.to_datetime(df[ 'ended_at'])
```

```python
df.info()
```

```python
df["trip_duration"] = (df["ended_at"] - df["started_at"]).dt.total_seconds()/60
df.head()
```

```python
df['trip_day'] = df['started_at'].dt.day_name()
df["trip_day"] = pd.Categorical(
    df["trip_day"],
    categories=['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'],
    ordered=True
)
df.head()
```

```python
sns.set_theme(style="whitegrid")
```

```python
plt.figure(figsize=(10, 5))
plot = sns.barplot(
    x='trip_day',
    y='trip_duration',
    data=df, 
    hue='member_casual', palette= {'member': '#74C365', 'casual': '#FF6347'},
    errorbar=None)
# Formatting title and labels
plt.title('Average Trip Duration by Day of Week', fontsize=14, fontweight='bold')
plt.xlabel('Day of the Week', fontsize=12)
plt.ylabel('Average Trip Duration(mins)', fontsize=12)
plt.legend(title='Member/Casual')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plot = sns.countplot(
    x='trip_day',
    data=df, 
    hue='member_casual', palette= {'member': '#74C365', 'casual': '#FF6347'}
    )

# Formatting title and labels
plt.title('Total Rides by Day of Week', fontsize=14, fontweight='bold')
plt.xlabel('Day of the Week', fontsize=12)
plt.ylabel('Number of Rides', fontsize=12)
plt.legend(title='Member/Casual')

plt.tight_layout()
plt.show()
```

```python
df['hour'] = df['started_at'].dt.hour
hourly_counts = df.groupby(['hour', 'member_casual']).size().reset_index(name='ride_count')
hourly_counts.head()
```

```python

plt.figure(figsize=(10, 5))
plot = sns.lineplot(
    x='hour', y='ride_count',
    data=hourly_counts, 
    hue='member_casual', palette= {'member': '#74C365', 'casual': '#FF6347'},
    marker='o',
    linewidth=2.2
   )
# Formatting title and labels
plt.title('Total Rides by Hour of Day', fontsize=14, fontweight='bold')
plt.xlabel('Hour of day(0-24)', fontsize=12)
plt.ylabel('Number of Rides', fontsize=12)
plt.legend(title='Member/Casual')

plt.tight_layout()
plt.show()
```
